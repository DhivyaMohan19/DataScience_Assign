```python
'''Exercise 1: Prime Numbers

Write a Python program that checks whether a given number is prime or not.
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.'''
```

```python
while True:
  user_input = int(input('Enter a number greater than 0: '))
  if user_input <= 0:
    print("Warning! Wrong input entered")
    continue
  else:
    break
factor = 0
for f in range (1, user_input+1):
  if (user_input % f  == 0):
    factor += 1
if factor <= 2:
  print(f'{user_input} is a prime number')
else:
  print(f'{user_input} is not Prime')
```

```python
'''Exercise 2: Product of Random Numbers
Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers.
The program should then check if the user's answer is correct and display an appropriate message.
'''
```

```python
import random

user_resp = '1'
while user_resp.upper() != 'E':
  num1 = random.randrange(0, 101)
  num2 = random.randrange(0, 101)

  user_resp = input(f"\n Can You Give the Product of {num1}  and {num2}? \n")

  if int(user_resp) == int(num1 * num2):
    print("Hurray!! You are Correct")
  else:
    print(f"You are Wrong :( !! the answer is {num1 * num2}")

  print("\n")
  user_resp = input("Please Enter 'E' to exit 'OR' press any key to Play Again: \n")
```

```python
'''

Exercise 4:
Problem Statement (Question): Hospital Billing System
A hospital charges its patients based on the following:
•	Room charges per day: ₹2,000
•	Doctor consultation fee (flat): ₹1,500
•	Lab test charges: Based on the number of tests taken. Each test costs ₹300
•	Medicine charges: Total cost of medicines provided
•	Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied
Write a Python program to calculate the final bill for a patient who:
•	Stayed for 4 days
•	Had 3 lab tests
•	Had medicine charges of ₹2,400
'''
```

```python
class hospital:

  room_chg = 2000
  doc_fee = 1500
  lab_chg = 300

  def __init__(self,days_cnt,tests_cnt,med_chgs):
      self.days_cnt = days_cnt
      self.tests_cnt = tests_cnt
      self.med_chgs = med_chgs


  def total_bill(self):
    total_bill = (self.room_chg * self.days_cnt) + (self.lab_chg * self.tests_cnt) + self.med_chgs + (self.doc_fee * self.days_cnt)

    if total_bill > 10000:
      print(f"Bill Amount is = {total_bill}, so Eligible for  10% Discount\n")
      total_bill = total_bill  - (total_bill * (10/100))
    return total_bill


# Creating an object of the hospital class

ndays= int(input("Enter the Number of Days Patients Stayed in the Hospital: "))
ntests = int(input("Enter the Number of Tests: "))
medicines = int(input("Enter the medicines purchase Amount: "))

patient = hospital(ndays,ntests,medicines) #creating instance of the object

#Calculating Total Bill Amount
print(f"The final bill for the patient who stayed for {ndays} days,took {ntests}lab tests and bought medicines worth {medicines} is {patient.total_bill()}")
```
